### Exploratory data analysis using forest-fires dataset

### Objective 1: Understanding which features influence fire area most

```python
sns.pairplot(df[['temp', 'RH', 'wind', 'rain', 'area']])
plt.suptitle("Key Features vs Area", y=1.02)
plt.show()
```

### objective2: Identifying Outliers Across Weather Features Using Boxplots

```python
features_to_melt = ['temp', 'RH', 'wind', 'rain', 'area']
melted_df = pd.melt(df[features_to_melt], var_name='Feature', value_name='Value')


plt.figure(figsize=(12, 6))
sns.boxplot(x='Feature', y='Value', hue='Feature', data=melted_df, palette='Set3', dodge=False, legend=False)

plt.yscale('log')
plt.title("Boxplot of Weather Features (Log Scale for Area)", fontsize=14)
plt.xlabel("Feature")
plt.ylabel("Value (log scale)")
plt.tight_layout()
plt.show()
```

```python

def remove_outliers_iqr(df, columns):
    clean_df = df.copy()
    for col in columns:
        Q1 = clean_df[col].quantile(0.25)
        Q3 = clean_df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        clean_df = clean_df[(clean_df[col] >= lower_bound) & (clean_df[col] <= upper_bound)]
    return clean_df
features_to_clean = ['temp', 'RH', 'wind', 'rain', 'area']

# Step 3: Apply the function to remove outliers
df_cleaned = remove_outliers_iqr(df, features_to_clean)

# Step 4: Melt the cleaned DataFrame for plotting
melted_df_clean = pd.melt(df_cleaned[features_to_clean], var_name='Feature', value_name='Value')

# Step 5: Boxplot without outliers
plt.figure(figsize=(12, 6))
sns.boxplot(x='Feature', y='Value', hue='Feature', data=melted_df_clean, palette='Set3', dodge=False, legend=False)

plt.yscale('log')  # Helps visualize wide-ranging values like 'area'
plt.title("Boxplot of Weather Features After Removing Outliers (Log Scale for Area)", fontsize=14)
plt.xlabel("Feature")
plt.ylabel("Value (log scale)")
plt.tight_layout()
plt.show()
```

### objective 3: Analyzing Distributions of Weather and Area Features

```python
fig, axes = plt.subplots(2, 3, figsize=(14, 8))
features = ['temp', 'RH', 'wind', 'rain', 'area']

for i, col in enumerate(features):
    sns.histplot(df[col], kde=True, ax=axes[i//3, i%3], bins=30)
    axes[i//3, i%3].set_title(f'Distribution of {col}')

plt.tight_layout()
plt.show()
```

### objective 4: Identifying Linear Relationships via Correlation Heatmap

```python
numeric_df = df.select_dtypes(include='number')
correlation_matrix = numeric_df.corr()

print(correlation_matrix)  # Check output

plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

### objective 5: Relationship between wind and fire size

```python
sns.scatterplot(data=df, x='wind', y='area')
plt.title("Wind Speed vs Fire Area")
plt.show()
```

### objective 6: Correlation Analysis of Features with Burned Area

```python
# Keep only numeric columns for correlation analysis
numeric_df = df.select_dtypes(include='number')

# Compute correlation matrix
correlation = numeric_df.corr()

# Show features most correlated with 'area'
print("Top Correlated Features with 'area':")
print(correlation['area'].sort_values(ascending=False))
```

### objective 7 : Visualize the Proportion of Fires by Day of the Week

```python
# Check if 'day' column exists in your dataset
if 'day' in df.columns:
    day_counts = df['day'].value_counts()

    # Pie chart with smaller size
    plt.figure(figsize=(5, 5))  # reduced from (8, 8) to (5, 5)
    plt.pie(day_counts, labels=day_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
    plt.title('Proportion of Fires by Day of the Week', fontsize=12)
    plt.axis('equal')  # Ensures pie is a circle
    plt.tight_layout()
    plt.show()
else:
    print("The 'day' column is not present in the dataset.")
```
